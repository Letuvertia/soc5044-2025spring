# 1. Application Programming Interfaces (APIs)

## What is an API? 
An API (Application Programming Interfaces) allows you to programmatically extract and interact with data under the hood of websites like Genius as well as other social networks, applications, and projects that make their data publicly available, such as Twitter and [The Smithsonian museums](https://www.si.edu/openaccess/devtools).

An API is something that a project or company explicitly designs for data-sharing purposes. Why do companies or projects go to all this trouble? One reason is that it helps to promote the use and further development of an application and its data. For example, Twitter wants other developers to use, integrate, and build upon Twitter tools and data. The Twitter API is the main conduit by which these developers can do so.

## Pros
Because APIs are explicitly designed for data-sharing purposes, working with an API is often a cleaner, more reliable, and more streamlined process than other ways of collecting internet data such as web scraping.

## Cons

One of the downsides is that the companies and projects that design the APIs get to decide exactly which kinds of data they want to share. \*Spoiler alert\* They often choose not to share their most lucrative and desirable data. For example, the Genius API does not provide access to lyrics data, and the Twitter API does not provide free access to tweets more than 14 days in the past. To get that Twitter data, you need to *pay* for special API access.

## Making an API Request

Making an API request looks a lot like typing a specially-formatted URL. But instead of getting a rendered HTML web page in return, you get some data in return.

There are a few different ways that we can query the Genius API, all of which are discussed in the [Genius API documentation](https://docs.genius.com/#songs-h2). The way we're going to cover in this lesson is [the basic search](https://docs.genius.com/#search-h2), which allows you to get a bunch of Genius data about any artist or songs that you search for:

`http://api.genius.com/search?q={search_term}&access_token={client_access_token}`

Sticking with our Missy Elliott theme/obsession, we're going to search for Genius data about Missy Elliott.

First we're going to assign the string "Missy Elliott" to the variable `search_term`. Then we're going to make an f-string URL that contains the variables `search_term` and `client_access_token`. Note that if your `search_term` contains blanks, it has be replaced with the unicode of blank `%20` to make the URL valid.

```python
search_term = "Missy Elliott"
```

```python
genius_search_url = f"http://api.genius.com/search?q={search_term}&access_token={client_access_token}"
genius_search_url = genius_search_url.replace(" ", "%20")
```

This URL is basically all we need to make a Genius API request. Want proof? Run the cell below and print this URL, then copy and paste it into a new tab in your web browser.

```python
print(genius_search_url)
```

It doesn't look pretty, but that's a bunch of Genius data about Missy Elliott!

We can programmatically do the same thing by again using the Python library `requests` with this URL. Instead of getting the `.text` of the response, as we did before, we're going to use `.json()`.

[JSON](https://www.w3schools.com/whatis/whatis_json.asp) is a data format that is commonly used by APIs. JSON data can be nested and contains key/value pairs, much like a Python dictionary.

```python
import requests
```

```python
response = requests.get(genius_search_url)
json_data = response.json()
```

The JSON data that we get from our Missy Elliott API query looks something like this:

```python
json_data
```

We can index this data (again, like a Python dictionary) and look at the first "hit" about Missy Elliott from Genius.com.

```python
json_data['response']['hits'][0]
```

We can tell that this data describes the song "Work It" and contains other information about the song, such as its number of Genius annotations, its number of web page views, and links to images of its album cover.

## Looping Through JSON Data

### Get Song Titles

```python
for song in json_data['response']['hits']:
    print(song['result']['full_title'])
```

### Get Song Tiles and Page View Counts

```python
for song in json_data['response']['hits']:
    print(song['result']['full_title'], song['result']['stats']['pageviews'])
```

### Transform Song Titles and Page View Counts into a DataFrame

We can loop through this data, append it into a list, and then transform that list into a Pandas dataframe by calling `pd.DataFrame()`

```python
import pandas as pd
```

:::{admonition} Pandas Review
:class: pandasreview
 Do you need a refresher or introduction to the Python data analysis library Pandas? Be sure to check out <a href="https://melaniewalsh.github.io/Intro-Cultural-Analytics/Data-Analysis/Pandas-Basics-Part1.html"> Pandas Basics (1-3) </a> in this textbook!
    
:::

```python
missy_songs = []
for song in json_data['response']['hits']:
    missy_songs.append([song['result']['full_title'], song['result']['stats']['pageviews']])
    
#Make a Pandas dataframe from a list
missy_df = pd.DataFrame(missy_songs)
missy_df.columns = ['song_title', 'page_views']
missy_df
```

## Transform Song Titles, Page View Counts, & Album Covers into a DataFrame

Just for fun, we can do the same thing but also add links to images of Missy Elliott's album art—and we can actually display those images, too!

To display images in a Pandas dataframe, you need to run `from IPython.core.display import HTML` and make the function `get_image_html()`. We're going to take the image URLs and make them into HTML objects.

```python
from IPython.core.display import HTML
```

```python
def get_image_html(link):
    image_html = f"<img src='{link}' width='500px'>"
    return image_html
```

```python
missy_songs = []
for song in json_data['response']['hits']:
    missy_songs.append([song['result']['full_title'], song['result']['stats']['pageviews'], song['result']['song_art_image_url']])
    
missy_df = pd.DataFrame(missy_songs)
missy_df.columns = ['song_title', 'page_views','album_cover_url']

#Use the function get_image_html()
missy_df['album_cover'] = missy_df['album_cover_url'].apply(get_image_html)
missy_df
```

If we call `HTML()` on our dataframe and add the method `.to_html(escape=False)` to the dataframe, then it should display the dataframe with viewable images.

```python
HTML(missy_df[['album_cover', 'page_views', 'song_title']].to_html(escape=False))
```

## `Get Artist`

The "Get Artist" endpoint requests information about an artist.

According to the [API reference](https://developer.spotify.com/documentation/web-api/reference/get-an-artist):
- Endpoint URL: `https://api.spotify.com/v1/artists/{id}`
- Arguments:
    - `id`: the Spotify ID of the artist.

We will stick to the Missy Elliott example.

An easy way to get the Spotify ID of an artist is using the Spotify Desktop App:

1. Search the artist
1. Click on the three dots icon from the artist profile
1. Select Share > Copy link to artist. The Spotify ID is the value that comes right after the open.spotify.com/artist URI.

```python
# Missy Elliott's Spotify artist ID
artist_id = "2wIVse2owClT7go1WT98tk"

# Send a GET request to the Spotify API
url = f"https://api.spotify.com/v1/artists/{artist_id}"
headers = {
    "Authorization": "Bearer " + access_token_json['access_token']
}
response = requests.get(url, headers=headers)
artist_data = response.json()
artist_data
```

## `Get Playlist`

The "Get Playlist" endpoint gets a playlist owned by a Spotify user.

According to the [API reference](https://developer.spotify.com/documentation/web-api/reference/get-playlist):
- Endpoint URL: `https://api.spotify.com/v1/playlists/{playlist_id}`
- Arguments:
    - `playlist_id`: the Spotify ID of the playlist.

2024/11/27 Update: According to [Spotify](https://developer.spotify.com/blog/2024-11-27-changes-to-the-web-api), Web API will no longer be able to access "Algorithmic and Spotify-owned editorial playlists", but only user-owned playlists.

```python
# "上班播放清單" playlist
# https://open.spotify.com/playlist/39oYWk8IhN8efflXzJWlHP
playlist_id = "39oYWk8IhN8efflXzJWlHP"

# Send a GET request to the Spotify API
url = f"https://api.spotify.com/v1/playlists/{playlist_id}"
headers = {
    "Authorization": "Bearer " + access_token_json['access_token']
}
response = requests.get(url, headers=headers)
playlist_data = response.json()
playlist_data
```

We can, for example, list the artist and the *popularity* of each song in the playlist. According to the [API reference](https://developer.spotify.com/documentation/web-api/reference/get-playlist) (scroll down to the "Response" section), the definition fo *popularity* is as follows:

> The popularity of the track. The value will be between 0 and 100, with 100 being the most popular.
> The popularity of a track is a value between 0 and 100, with 100 being the most popular. The popularity is calculated by algorithm and is based, in the most part, on the total number of plays the track has had and how recent those plays are.
> Generally speaking, songs that are being played a lot now will have a higher popularity than songs that were played a lot in the past.

We can also use the artist id in the response to request again with the `Get Artist` endpoint to get the number of followers of the artists.

```python
def get_artist_followers(artist_id):
    # Send a GET request to the Spotify API
    url = f"https://api.spotify.com/v1/artists/{artist_id}"
    headers = {
        "Authorization": "Bearer " + access_token_json['access_token']
    }
    response = requests.get(url, headers=headers)
    artist_data = response.json()
    return artist_data['followers']['total']


playlist_tracks = []
for track in playlist_data['tracks']['items']:
    track_name = track['track']['name']
    track_artist = track['track']['artists'][0]['name']
    track_popularity = track['track']['popularity']
    track_artist_followers = get_artist_followers(track['track']['artists'][0]['id'])
    playlist_tracks.append([track_name, track_artist, track_artist_followers, track_popularity])

playlist_df = pd.DataFrame(playlist_tracks)
playlist_df.columns = ['track_name', 'track_artist', 'track_artist_followers', 'track_popularity']
playlist_df
```
